```python
%python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType

schema = StructType([
    StructField("col1", StringType(), True),
    StructField("col2", IntegerType(), True),
    # Add all columns as per your CSV file
])

(
    spark.readStream.format("csv")
    .schema(schema)
    .load("s3://sawfeproject1/silver")
    .writeStream.option("checkpointLocation", "s3://sawfeproject1/pradeep/checkpoint")
    .trigger(availableNow=True)
    .toTable("pradeep.default.pradeep12")
)
```
